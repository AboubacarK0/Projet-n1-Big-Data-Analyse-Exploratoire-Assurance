Il y a eu des soucis d'importations du dataframe dossier, il nous a fallu utiliser une méthode d'importation plus robuste

```python
# Parseur personnalisé pour recréer le DataFrame dossier.csv from scratch
import csv
from io import StringIO

def parse_dossier_csv(file_path):
    rows = []
    with open(file_path, 'r', encoding='latin1') as f:
        reader = csv.reader(f, quoting=csv.QUOTE_ALL)
        for row in reader:
            # row est ['ligne entière'], parser la ligne entière
            line = row[0]
            # Enlever les guillemets externes
            if line.startswith('"') and line.endswith('"'):
                line = line[1:-1]
            # Maintenant parser avec csv
            sub_reader = csv.reader(StringIO(line), delimiter=',', quotechar='"', doublequote=True)
            parsed_row = next(sub_reader)
            rows.append(parsed_row)
    
    # La première ligne est l'en-tête
    header = rows[0]
    data = rows[1:]
    return pd.DataFrame(data, columns=header)

# Utiliser le parseur
dossier = parse_dossier_csv("dossier.csv")
print("DataFrame recréé from scratch")
# Affichage des dimensions de chaque base
print("Dimensions de df_dossier :", dossier.shape)
print("Dimensions de df_ressources :", ressources.shape)
print("Dimensions de df_temps :", temps.shape)
```

## 2. Traitement de données

Remarques sur l'état initial des bases données:  
- Les datasets qui viennent d'être importés présentent des incohérences au niveau des format des colonnes (Dtype). Pour la suite de l'analyse, il sera pertienent de changer les formats des colonnes.  Les colonnes de date seront mises sous le bon format date pour permettre de faire des analyses temporelles
- Les mots dans les noms de colonnes sont séparés par des points. Pour éviter des soucis d'erreur de syntaxte python, les points seront remplacés par des tirets  
- Des informations manquent dans certaines cellules. Exemple: Des cellules contiennent '???', ces cellules seront désormais considérées comme des Nan


```python

# 1. Remplacement des points par des tirets du bas dans les noms de colonnes
dossier.columns = dossier.columns.str.replace('.', '_', regex=False)
ressources.columns = ressources.columns.str.replace('.', '_', regex=False)
temps.columns = temps.columns.str.replace('.', '_', regex=False)


# 2. Remplacement des valeurs aberrantes connues par NaN
dossier.replace("???", np.nan, inplace=True)
ressources.replace("???", np.nan, inplace=True)
temps.replace("???", np.nan, inplace=True)
```

### 2.1. Le DataFrame Dossier

```python
dossier.head()
```

```python
# 1. Conversion des types pour le DataFrame 'dossier'
dossier['date_de_survenance'] = pd.to_datetime(dossier['date_de_survenance'], format='%Y/%m/%d', errors='coerce')
# Identifiants et indicateurs en entiers
colonnes_numeriques_dossier = [
     'TOP_D_R', 'TOP_VR', 
    'TOP_Rappat_valide', 'TOP_Poursuite', 'TOP_Recup', 'TOP_Autres_Garanties'
]
for col in colonnes_numeriques_dossier:
    dossier[col] = pd.to_numeric(dossier[col], errors='coerce').astype('Int64')
```

```python
# Fusion de la date et de l'heure pour le DataFrame 'dossier'
# Conversion préalable de date_ouverture en datetime
dossier['date_ouverture'] = pd.to_datetime(dossier['date_ouverture'], format='%Y/%m/%d', errors='coerce')

# Création puis conversion en datetime
combinaison_dossier = dossier['date_ouverture'].dt.strftime('%Y-%m-%d') + ' ' + dossier['heure_ouverture']
dossier['datetime_ouverture'] = pd.to_datetime(combinaison_dossier, format='%Y-%m-%d %H:%M:%S', errors='coerce')

# Suppression des anciennes colonnes devenues redondantes
dossier.drop(columns=['date_ouverture', 'heure_ouverture'], inplace=True)

# Vérification des nouveaux types et aperçu des résultats
print("--- Structure mise à jour pour 'dossier' ---")
print(dossier[['datetime_ouverture']].dtypes)
print(dossier['datetime_ouverture'].head(2))
```

Traitement des doublons et des cellules vides 

```python
#  Identification et décompte des doublons parfaits (lignes 100% identiques)
nb_doublons_complets = dossier.duplicated().sum()
print(f"Nombre de lignes doublons (exactes) : {nb_doublons_complets}")

# Identification des doublons sur l'identifiant unique 'Numero_dossier_ID'
nb_doublons_id = dossier.duplicated(subset=['Numero_dossier_ID']).sum()
print(f"Nombre de doublons basés uniquement sur Numero_dossier_ID : {nb_doublons_id}")

# Calcul du pourcentage de valeurs manquantes par colonne
Total_manquants = dossier.isna().sum()
Pourcentage_manquants = (dossier.isna().sum() / len(dossier)) * 100

bilan_manquants = pd.DataFrame({
    'Valeurs Manquantes': Total_manquants,
    'Pourcentage (%)': Pourcentage_manquants
})

print("\n--- Bilan des valeurs manquantes dans 'dossier' ---")
print(bilan_manquants[bilan_manquants['Valeurs Manquantes'] > 0])
```

```python
# Suppression des doublons parfaits (lignes 100% identiques)
dossier.drop_duplicates(inplace=True)
print(f"Dimensions après suppression des doublons parfaits : {dossier.shape}")
```

```python
extraits_doublons = dossier[dossier.duplicated(subset=['Numero_dossier_ID'], keep=False)].sort_values(by='Numero_dossier_ID')

print("\n--- Extrait des dossiers ayant plusieurs lignes distinctes ---")
# Affichage des 6 premières lignes du sous-ensemble pour inspection visuelle
print(extraits_doublons[['Numero_dossier_ID', 'Matricule_de_traitement', 'datetime_ouverture', 'Cause_intervention']].head(6))
```

Puisque l'objectif final inclut des modélisations économétriques où chaque observation doit représenter un dossier unique et complet, nous avons juger pertinent de filtrer ces doublons pour ne conserver que la ligne la plus informative (celle contenant le moins de valeurs manquantes) ou, à défaut, la première occurrence.

```python
# Calcul du nombre de valeurs non manquantes par ligne pour prioriser les lignes les plus complètes
dossier['nb_valeurs_reelles'] = dossier.notna().sum(axis=1)

# Tri décroissant selon la complétude, puis conservation de la meilleure ligne pour chaque ID
dossier.sort_values(by='nb_valeurs_reelles', ascending=False, inplace=True)
dossier.drop_duplicates(subset=['Numero_dossier_ID'], keep='first', inplace=True)

# Suppression de la colonne temporaire
dossier.drop(columns=['nb_valeurs_reelles'], inplace=True)


# Imputation des variables catégorielles manquantes par la modalité "Inconnu"
colonnes_categorieles = ['Formule', 'Assistance_ou_Administratif']
for col in colonnes_categorieles:
    dossier[col] = dossier[col].fillna("Inconnu")


# 3. Suppression des observations sans repère temporel valide
dossier.dropna(subset=['date_de_survenance', 'datetime_ouverture'], inplace=True)


# 4. Bilan final du nettoyage de la table dossier
print(f"Dimensions finales du DataFrame 'dossier' : {dossier.shape}")
print("\n--- Décompte final des valeurs manquantes ---")
print(dossier.isna().sum())
```

### 2.2. Le DataFrame ressources 

```python
# 1. Décompte des doublons parfaits (lignes 100% identiques)
nb_doublons_ressources = ressources.duplicated().sum()
print(f"Nombre de lignes doublons (exactes) dans ressources : {nb_doublons_ressources}")

# 2. Vérification de l'unicité du couple Clé Primaire (Matricule + Date_presence)
# Un collaborateur ne devrait idéalement figurer qu'une seule fois par jour de présence
nb_doublons_cle = ressources.duplicated(subset=['Matricule', 'Date_presence']).sum()
print(f"Nombre de doublons sur le couple (Matricule, Date_presence) : {nb_doublons_cle}")

# 3. Décompte et pourcentage de valeurs manquantes par colonne
manquants_ressources = ressources.isna().sum()
pourcentage_ressources = (ressources.isna().sum() / len(ressources)) * 100

bilan_ressources = pd.DataFrame({
    'Valeurs Manquantes': manquants_ressources,
    'Pourcentage (%)': pourcentage_ressources
})

print("\n--- Bilan des valeurs manquantes dans 'ressources' ---")
print(bilan_ressources)
```

```python
ressources['Matricule'] = ressources['Matricule'].astype(str).str.strip()   
ressources.info()
```

Les résultats indiquent que le DataFrame ressources est parfaitement intègre. L'absence de doublons complets, l'unicité du couple clé primaire (Matricule, Date_presence), ainsi que l'inexistance totale de valeurs manquantes permettent de valider directement cette table sans aucune action de nettoyage supplémentaire.

La prochaine étape logique du traitement consiste à analyser et nettoyer la troisième et dernière base de données : temps.

### 2.3. Le DataFrame TEMPS

```python
# 1. Décompte des doublons parfaits (lignes 100% identiques)
nb_doublons_temps = temps.duplicated().sum()
print(f"Nombre de lignes doublons (exactes) dans temps : {nb_doublons_temps}")

# 2. Vérification de la présence de doublons sur la clé d'identification
# Une ligne représente une phase de traitement. Plusieurs lignes peuvent exister pour un même dossier 
# s'il y a eu plusieurs étapes ou plusieurs intervenants (Matricule).
# On vérifie ici si un même collaborateur possède plusieurs entrées identiques pour le même dossier au même instant précis.
nb_doublons_phase = temps.duplicated(subset=[
    'Numero_dossier', 'Matricule', 'Date_debut_traitement', 'heure_debut_traitement'
]).sum()
print(f"Nombre de doublons sur le triplet (Numero_dossier, Matricule, datetime_debut_traitement) : {nb_doublons_phase}")

# 3. Décompte et pourcentage de valeurs manquantes par colonne
manquants_temps = temps.isna().sum()
pourcentage_temps = (temps.isna().sum() / len(temps)) * 100

bilan_temps = pd.DataFrame({
    'Valeurs Manquantes': manquants_temps,
    'Pourcentage (%)': pourcentage_temps
})

print("\n--- Bilan des valeurs manquantes dans 'temps' ---")
print(bilan_temps)
```

```python
# Fusion de la date et de l'heure en un objet datetime complet
temps['datetime_debut_traitement'] = pd.to_datetime(
    temps['Date_debut_traitement'].astype(str) + ' ' + temps['heure_debut_traitement'].astype(str),
    format='%Y/%m/%d %H:%M:%S',
    errors='coerce'
)
temps.drop(columns=['Date_debut_traitement', 'heure_debut_traitement'], inplace=True)

# Nettoyage (duree_corrigee)
temps.dropna(subset=['duree_corrigee'], inplace=True)

# Vérification finale de la structure et des valeurs manquantes de la table 'temps'
print("--- Structure finale du DataFrame 'temps' ---")
print(temps.dtypes)

print("\n--- Décompte final des valeurs manquantes ---")
print(temps.isna().sum())

print(f"\nDimensions finales de la table 'temps' : {temps.shape}")
```

Le numéro de dossier et le numéro de matricule sont transformés en format string pour les futures jointures et une bonne harmonisation.

```python
temps['Numero_dossier'] = temps['Numero_dossier'].astype(str).str.strip()
temps['Matricule'] = temps['Matricule'].astype(str).str.strip()
temps.info()
```

```python
temps.head()
```

```python
# 1. changement des types de données pour la table 'ressources' et 'temps'
# Ajustement des derniers types dans la table 'ressources'
# Conversion de la date (qui était en JJ/MM/AAAA) pour correspondre au format datetime
ressources['Date_presence'] = pd.to_datetime(ressources['Date_presence'], format='%d/%m/%Y', errors='coerce')
# Conversion de Duree_travail en décimal (float) pour l'économie
ressources['Duree_travail'] = pd.to_numeric(ressources['Duree_travail'], errors='coerce')

# . Préparation de la clé de date dans 'temps'
temps['Date_presence_cle'] = pd.to_datetime(temps['datetime_debut_traitement'].dt.date)

# Pour une jointure parfaite, on s'assure que les deux colonnes de date ont exactement le même type normalisé
temps['Date_presence_cle'] = pd.to_datetime(temps['Date_presence_cle']).dt.normalize()
ressources['Date_presence'] = ressources['Date_presence'].dt.normalize()


# 2. Première fusion : Table 'temps' + Table 'dossier'
df_fusion = pd.merge(
    temps, 
    dossier, 
    left_on='Numero_dossier', 
    right_on='Numero_dossier_ID', 
    how='inner'
)


# 3. Deuxième fusion : Base fusionnée + Table 'ressources'
df_final = pd.merge(
    df_fusion, 
    ressources, 
    left_on=['Matricule', 'Date_presence_cle'], 
    right_on=['Matricule', 'Date_presence'], 
    how='inner'
)


# 4. Nettoyage des colonnes clés devenues redondantes après la jointure
colonnes_a_supprimer = ['Numero_dossier_ID', 'Date_presence_cle', 'Date_presence']
df_final.drop(columns=colonnes_a_supprimer, inplace=True, errors='ignore')


# 5. Exportation du fichier final propre pour un nouveau notebook dédié à la partie économétrie
df_final.to_csv("base_econometrie.csv", index=False)


print(f"Dimensions de la base finale : {df_final.shape}")
```
